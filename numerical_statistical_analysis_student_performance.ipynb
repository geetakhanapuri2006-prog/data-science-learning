# Basic Statistical Analysis of Numerical Attributes

**Objective:** Extract numerical attributes from a dataset and perform basic statistical analysis using **NumPy** and **Pandas**.

Measures: Mean, Median, Mode, Standard Deviation, Minimum, Maximum, and Range.

**Learning Outcome:** Perform and interpret basic statistical analysis of numerical data.

## 1. Dataset: Student Performance

This self-contained dataset represents student academic performance. Numerical attributes include age, study hours, attendance, assignment score, midterm score, final score, and previous GPA.

```python
import numpy as np
import pandas as pd

print('NumPy:', np.__version__)
print('Pandas:', pd.__version__)
```

```python
data = {
    'Age': [18, 19, 20, 18, 21, 19, 22, 20, 18, 23, 21, 19, 20, 22, 18, 21, 20, 19, 22, 21],
    'Study_Hours': [2.5, 3.0, 4.0, 2.0, 5.0, 3.5, 6.0, 4.5, 2.5, 5.5, 4.0, 3.0, 4.5, 6.0, 2.0, 5.0, 4.0, 3.5, 5.5, 4.5],
    'Attendance': [72, 78, 85, 68, 92, 80, 95, 88, 74, 90, 86, 79, 84, 93, 70, 91, 87, 81, 94, 89],
    'Assignments_Score': [65, 72, 80, 60, 88, 75, 92, 85, 68, 90, 82, 73, 79, 94, 62, 89, 84, 76, 91, 86],
    'Midterm_Score': [58, 65, 74, 55, 82, 70, 88, 79, 61, 85, 77, 67, 73, 90, 57, 84, 80, 71, 87, 81],
    'Final_Score': [62, 68, 78, 59, 86, 73, 91, 82, 65, 88, 80, 70, 76, 93, 61, 87, 83, 74, 90, 85],
    'Previous_GPA': [6.2, 6.8, 7.4, 5.9, 8.1, 7.0, 8.8, 7.9, 6.4, 8.5, 7.7, 6.7, 7.3, 9.0, 6.0, 8.3, 8.0, 7.1, 8.7, 8.2],
}

df = pd.DataFrame(data)
df.to_csv('student_performance.csv', index=False)

print('Dataset shape:', df.shape)
display(df.head())
```

## 2. Inspect the Dataset

```python
print('Dataset information:')
df.info()

print('\nMissing values:')
display(df.isnull().sum())
```

## 3. Extract Numerical Attributes

```python
numeric_data = df.select_dtypes(include=np.number)
print('Numerical attributes:')
print(list(numeric_data.columns))
```

## 4. Central Tendency: Mean, Median and Mode

```python
central_tendency = pd.DataFrame({
    'Mean': numeric_data.mean(),
    'Median': numeric_data.median(),
    'Mode': numeric_data.mode().iloc[0]
})
display(central_tendency.round(2))
```

**Interpretation:** Mean is the average, median is the middle observation, and mode is the most frequently occurring value. Comparing mean and median can provide an initial indication of skewness.

## 5. Dispersion: Standard Deviation, Minimum and Maximum

```python
dispersion = pd.DataFrame({
    'Standard Deviation': numeric_data.std(),
    'Minimum': numeric_data.min(),
    'Maximum': numeric_data.max(),
    'Range': numeric_data.max() - numeric_data.min()
})
display(dispersion.round(2))
```

**Interpretation:** Standard deviation measures spread around the mean. Larger values indicate greater variability. Minimum and maximum show the observed limits, while range is maximum minus minimum.

## 6. Complete Statistical Summary

```python
statistical_summary = pd.DataFrame({
    'Mean': numeric_data.mean(),
    'Median': numeric_data.median(),
    'Mode': numeric_data.mode().iloc[0],
    'Standard Deviation': numeric_data.std(),
    'Minimum': numeric_data.min(),
    'Maximum': numeric_data.max()
})
display(statistical_summary.round(2))
```

## 7. NumPy Analysis of Final Score

```python
values = numeric_data['Final_Score'].to_numpy()

print('Final Score - NumPy analysis')
print('Mean              :', round(np.mean(values), 2))
print('Median            :', round(np.median(values), 2))
print('Standard deviation:', round(np.std(values), 2))
print('Minimum           :', np.min(values))
print('Maximum           :', np.max(values))
```

## 8. Interpretation of Results

1. **Age:** The age range is relatively narrow, indicating a fairly homogeneous student group.
2. **Study Hours:** Variation indicates different study patterns among students.
3. **Attendance:** The mean and spread indicate the typical attendance level and how much students differ.
4. **Academic Scores:** Assignment, midterm, and final scores describe academic performance and its variability.
5. **Previous GPA:** Provides an indicator of prior academic performance.
6. **Overall:** Mean and median describe central tendency, while standard deviation and range describe dispersion. These measures together provide a basic understanding of the distribution and characteristics of numerical data.
